# [Dataset 2](https://www.kaggle.com/datasets/pashupatigupta/emotion-detection-from-text)

```python
df2 = pd.read_csv("../data/raw/extra_dataset_PashupatiGupta.csv")
df2.head()
```

```python
df2 = df2.drop('tweet_id', axis=1)
df2 = df2.rename(columns={'sentiment': 'label', 'content': 'text'})
df2.head()
```

```python
df2 = df2[df2['label'].isin(['love', 'surprise', 'enthusiasm'])]
df2.head()
```

```python
df2['label'].value_counts()
```

```python
df2['label'] = df2['label'].replace('enthusiasm', 'surprise')
df2['label'].value_counts()
```

## Merging df1 and df2...

```python
df_merged = pd.concat([df1, df2], ignore_index=True)
df_merged.head()
```

```python
df_merged['label'].value_counts()
```

# [Dataset 4](https://www.kaggle.com/datasets/bryanhuerta/social-media-sentiment-data)

```python
df4 = pd.read_csv("../data/raw/extra_dataset_BryanHuerta.csv")
df4.head()
```

```python
df4 = df4[['Text', 'Sentiment']]
df4 = df4.rename(columns={'Text': 'text', 'Sentiment': 'label'})
```

```python
unique_labels = df4['label'].unique()
print(', '.join(unique_labels))
```

Awe, Gratitude, Love, Confusion, Arousal, Adoration, Admiration, Thrill, Adrenaline, Wonder, Energy, Suspense


```python
labels_to_keep = ['Awe', 'Gratitude', 'Love', 'Confusion', 'Arousal', 'Adoration', 'Admiration', 'Thrill', 'Adrenaline', 'Wonder', 'Energy', 'Suspense']
df4_copy = df4[df4['label'].isin(labels_to_keep)].copy()
df4_copy['label'].value_counts()
```

```python
labels_to_display = ['Wonder', 'Adrenaline', 'Arousal', 'Thrill', 'Suspense']
filtered_df = df4_copy[df4_copy['label'].isin(labels_to_display)]
for index, row in filtered_df.iterrows():
    print(f"{row['text']}, {row['label']}")
```

```python
filtered_df = df4_copy[df4_copy['label'].isin(['Awe'])]
for index, row in filtered_df.iterrows():
    print(f"{row['text']}, {row['label']}")
```

```python
labels_to_surprise = ['Surprise', 'Awe', 'Confusion', 'Arousal', 'Thrill', 'Adrenaline', 'Wonder', 'Energy', 'Suspense']
labels_to_love = ['Gratitude', 'Love', 'Adoration', 'Admiration']
```

```python
df4_copy['label'] = df4_copy['label'].replace(labels_to_surprise, 'surprise')
df4_copy['label'] = df4_copy['label'].replace(labels_to_love, 'love')
df4_copy['label'].value_counts()
```

```python
df_merged = pd.concat([df_merged, df4_copy], ignore_index=True)
df_merged['label'].value_counts()
```
